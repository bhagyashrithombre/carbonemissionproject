# Carbon Emission Forecasting System

## Objective

The objective of this project is to forecast future carbon emissions using historical industrial data and deep learning (LSTM).

The model analyzes energy consumption, production output, transportation, raw material usage, and environmental factors to predict future carbon emissions.

The system also helps industries understand emission trends and supports sustainability planning.

#### Feature Engineering

```python
target = "Carbon_Emission_tCO2e_TARGET"

features = [
    "Total_Energy_Consumption_kWh",
    "Renewable_Energy_Consumption_kWh",
    "NonRenewable_Energy_Consumption_kWh",
    "Production_Output_Units",
    "Supply_Chain_Transport_km",
    "Raw_Material_Usage_kg",
    "Energy_Cost_USD",
    "Carbon_Tax_USD",
    "Process_Efficiency_Percent",
    "Employment_Count",
    "Public_Acceptance_Index",
    "Strategy_Implementation_Cost_USD",
    "Expected_Carbon_Reduction_Percent",
    "Expected_Renewable_Share_Percent",
    "Social_Impact_Score"
]
```

```python
lstm_scaler = MinMaxScaler()
target_scaler = MinMaxScaler()

X_numeric_scaled = lstm_scaler.fit_transform(df[features])
y_scaled = target_scaler.fit_transform(df[[target]])

lstm_df = pd.DataFrame(X_numeric_scaled, columns=features)
lstm_df["Target"] = y_scaled
lstm_df["Company_ID"] = df["Company_ID"].values
```

```python
sequence_length = 60  # days of HISTORY in addition to the current day
X_lstm, y_lstm = [], []
for company in lstm_df["Company_ID"].unique():
    company_data = lstm_df[lstm_df["Company_ID"] == company].reset_index(drop=True)
    X_company = company_data.drop(["Company_ID", "Target"], axis=1).values
    y_company = company_data["Target"].values
    # Each sequence spans i-sequence_length+1 .. i INCLUSIVE, so the current day's
    # own features are the last timestep the model sees. Predicting y[i] from
    # only PAST days (excluding day i itself) was the main bug in the first
    # day i's features from the LSTM starved it of the most important input.
    for i in range(sequence_length - 1, len(company_data)):
        X_lstm.append(X_company[i - sequence_length + 1: i + 1])
        y_lstm.append(y_company[i])
X_lstm = np.array(X_lstm)
y_lstm = np.array(y_lstm)
print(X_lstm.shape)
print(y_lstm.shape)
```

```python
train_size = int(len(X_lstm) * 0.8)

X_train = X_lstm[:train_size]
X_test = X_lstm[train_size:]

y_train = y_lstm[:train_size].reshape(-1, 1)
y_test = y_lstm[train_size:].reshape(-1, 1)
```

```python
model = Sequential()

model.add(LSTM(128, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dropout(0.2))

model.add(LSTM(64))
model.add(Dropout(0.2))

model.add(Dense(32, activation="relu"))
model.add(Dense(1))

model.compile(optimizer="adam", loss="mse", metrics=["mae"])
model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_lr=1e-6
)

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_data=(X_test, y_test),
    callbacks=[early_stop, reduce_lr]
)
```

```python
plt.figure(figsize=(10,5))
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("LSTM Training vs Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE (scaled)")
plt.legend()
plt.show()
```

```python
lstm_prediction = model.predict(X_test)
lstm_prediction = target_scaler.inverse_transform(lstm_prediction)
actual = target_scaler.inverse_transform(y_test)

lstm_mae = mean_absolute_error(actual, lstm_prediction)
lstm_rmse = np.sqrt(mean_squared_error(actual, lstm_prediction))
lstm_r2 = r2_score(actual, lstm_prediction)

print("LSTM MAE :", lstm_mae)
print("LSTM RMSE:", lstm_rmse)
print("LSTM R2  :", lstm_r2)
```

```python
plt.figure(figsize=(15,5))
plt.plot(actual[:200], label="Actual")
plt.plot(lstm_prediction[:200], label="Predicted", linestyle="--")
plt.title("LSTM: Actual vs Predicted Carbon Emission (first 200 test points)")
plt.xlabel("Test sample index")
plt.ylabel("Carbon Emission (tCO2e)")
plt.legend()
plt.show()
```

```python
def simulate_reduction(reduction_pct, X_base, feature_names, model,
                        x_scaler, y_scaler,
                        feature_to_reduce="NonRenewable_Energy_Consumption_kWh"):
    """Reduce a feature by reduction_pct% across every timestep of every test sequence, then re-predict."""
    idx = feature_names.index(feature_to_reduce)

    # min and max of this ONE feature, as used by MinMaxScaler
    f_min = x_scaler.data_min_[idx]
    f_max = x_scaler.data_max_[idx]

    X_scenario = X_base.copy()

    # step 1: undo scaling -> get back real kWh values
    real_values = X_scenario[:, :, idx] * (f_max - f_min) + f_min

    # step 2: apply the % reduction on the REAL values (this is the fix)
    real_values = real_values * (1 - reduction_pct / 100.0)

    # step 3: redo scaling -> back to 0-1 range for the model
    X_scenario[:, :, idx] = (real_values - f_min) / (f_max - f_min)

    pred_scaled = model.predict(X_scenario, verbose=0)
    return y_scaler.inverse_transform(pred_scaled)


scenarios_pct = [0, 10, 20, 30]
scenario_results = {
    f"{pct}% reduction" if pct else "Current": simulate_reduction(
        pct, X_test, features, model, lstm_scaler, target_scaler
    ).mean()
    for pct in scenarios_pct
}

for name, val in scenario_results.items():
    print(f"{name}: {val:.2f} tCO2e (avg)")
```

```python
plt.figure(figsize=(7,5))
plt.bar(scenario_results.keys(), scenario_results.values(), color="seagreen")
plt.ylabel("Average Carbon Emission (tCO2e)")
plt.title("Scenario Simulation — Non-Renewable Energy Reduction")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
